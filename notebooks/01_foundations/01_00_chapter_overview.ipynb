## 3. The idea

### ❓ What is a language model?

Think of the children's game "hot and cold": something is hidden, you take a step, and a friend says "warmer" or "colder". You don't need to understand anything about the room to win. You just keep moving the way that gets you "warmer". A language model plays that game with text: it guesses what comes next, gets told how far off it was, and shifts its numbers a little so the next guess is warmer. Do that thousands of times and the guesses get good. The game makes one thing look easier than it is: you only move across a floor, and our model has 11,840 numbers to move at once, so it needs some maths to tell which way is warmer for each of them ([01_03](01_03_how_a_program_learns.ipynb)).

So, a language model is a program that reads some text and **guesses what comes next**. Give it `ἄνδρα μο` and it guesses the next character, `ι`, because `ἄνδρα μοι` ("tell me of the man") is how the Odyssey begins.

Ours guesses one character at a time. The big ones guess a **token**: a small piece of text, often part of a word. (Chapter 2 is all about tokens.)

### ❓ How can guessing the next character write a whole poem?

By doing it over and over. Guess one character, stick it on the end, then guess again with the longer text:

```
ἄνδρα μο → ι → ἄνδρα μοι → (space) → ἄνδρα μοι  → ἔ → …
```

Every chatbot works like this underneath. It never plans a whole sentence. It picks the next piece, then the next, and so on.

### ❓ What isn't a language model?

It isn't a database of facts, and it doesn't look anything up. It doesn't understand Greek the way you do, either. What it has is **patterns**: which characters tend to follow which, learned from the text it trained on. So ours knows Book 1 of the Odyssey and nothing else.

### ❓ What makes a model "large"?

The number of **weights**, the adjustable numbers inside it. More weights means room for more patterns.

| Model | Weights |
|---|---|
| ours, `HomerGreekLLM` | 11,840 |
| GPT-2 small (2019) | 124 million |
| today's big chat models | hundreds of billions |

Ours is tiny on purpose. It's the same idea at toy size, so it trains on a laptop in minutes. "Large" really just means "lots of weights".
