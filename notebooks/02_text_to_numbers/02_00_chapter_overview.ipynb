## 2. Words you'll meet

| Word | Meaning in plain English | Example |
|---|---|---|
| **token** | one piece of text the model reads as a single unit | `α`, or `λλα` |
| **vocabulary** | the list of all the tokens the model knows | our 137 tokens |
| **token ID** | the number a token gets in the vocabulary | `π` → 41 in the real Book 1 vocabulary (in 02_01's 5-character toy it's 4) |
| **encode** | turn text into token IDs | `πολλὰ` → `[41, 40, 36, 36, 110]` |
| **decode** | turn token IDs back into text | `[41, 40, 36, 36, 110]` → `πολλὰ` |
| **tokenizer** | the tool that does the encoding and decoding | `CharTokenizer` |
| **special token** | an extra token that is a signal, not real text | `<\|unk\|>`, `<\|endoftext\|>` |
| **BPE** | byte pair encoding: a way to make tokens out of common groups of letters | `ν` + space → one token |

## 3. The idea

### ❓ So how do we turn Greek into numbers?

Remember the secret code from school, where A is 1, B is 2, and a message becomes a row of numbers? This chapter builds that code for Homer's Greek: a key sheet with one number for every character, a way to write the poem as numbers, and a way to read it back without losing a letter. Then we cut the numbered poem into flashcards, so the model has something to practise on ([02_04](02_04_sliding_window.ipynb)). The code picture stops working at one point: in a secret code the numbers are only labels, but a model does maths with them, and then 41 would count as "more" than 40. That's why the chapter doesn't end with the code. It ends by swapping each number for a small list of numbers the model can adjust ([02_05](02_05_token_embeddings.ipynb)).

Building the code takes two steps:

1. Cut the text into small pieces, called **tokens**. The simplest choice is one letter = one token.
2. Give every different token its own number, its **token ID**. The list of all the tokens is the **vocabulary**.

```
πολλὰ  →  π  ο  λ  λ  ὰ  →  41  40  36  36  110
```

### ❓ Can we get the text back from the numbers?

Yes. Turning text into IDs is called **encoding**, and turning IDs back into text is decoding. If you encode and then decode, you should get back exactly what you started with. That check is called a round trip. And the tool that does both jobs is the tokenizer.

### ❓ What happens to a letter that isn't in the vocabulary?

The lazy option is to skip it. But then `Ζεὺς!` quietly becomes `Ζεὺς`, and nobody can tell that anything went missing. So we add a **special token**, `<|unk|>` ("unknown"), which stands in for any letter the tokenizer has never seen.

### ❓ How does the model know where one book ends and the next begins?

A line break won't do, because every verse already ends with one. So we put another special token, `<|endoftext|>`, between books.

### ❓ Do tokens have to be single letters?

No. **BPE** (byte pair encoding) builds bigger tokens out of groups of letters that turn up a lot, like `πολλ` ("many"). And bigger tokens mean the model sees more of the poem at once.

## 4. Recap

- A model only sees numbers: text is cut into tokens, and each token gets an ID from the vocabulary.
- Encode (text → IDs) and decode (IDs → text) must give back exactly the same text.
- Special tokens mark unknown characters and book ends; BPE makes bigger tokens from common groups.
