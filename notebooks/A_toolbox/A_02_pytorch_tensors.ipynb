## 3. The idea

### ❓ What's PyTorch?

**PyTorch** is a free library for doing maths on big lists of numbers, made for AI. It was first built at Meta (Facebook's company), and the PyTorch Foundation looks after it.

Most AI researchers and engineers use it, for two reasons. It's fast, and it can work out how to adjust the numbers so a model improves. That second part is what makes learning possible.

In code it's called `torch`.

## 4. Step by step

Each step is one question. Read it, guess what the code will do, run it, then look at what you got.

### ❓ What's a tensor?

A **tensor** is PyTorch's list of numbers. It looks like a Python list, but PyTorch can do maths on all of it at once, very fast.

```
πολλὰ   →   π    ο    λ    λ    ὰ
            41   40   36   36   110     one tensor of 5 numbers
```

👉 **Predict first:** what does `ids * 2` give?

````python
import torch

ids = torch.tensor([41, 40, 36, 36, 110])
print(ids)
print(ids * 2)
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

Every number doubled at once. With a plain list you'd need a loop. (These 5 numbers are `πολλὰ` in our tokenizer.)

</details>

### ❓ What's a shape?

The **shape** says how many numbers a tensor has in each direction. A row of 5 numbers has shape `(5,)`. A table with 2 rows and 3 columns has shape `(2, 3)`.

👉 **Predict first:** what shape does the table have?

````python
row = torch.tensor([41, 40, 36, 36, 110])
table = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(row.shape)
print(table.shape)
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`torch.Size([5])` and `torch.Size([2, 3])`.

</details>

### ❓ What's a dtype, and why does the code say `dtype=torch.long`?

The **dtype** (data type) is the kind of number stored. `torch.long` holds whole numbers, which is what token IDs are. There's no token number 4.5. `torch.float32` holds decimals, which is what the model's weights are.

`load_data` in `train_homer.py` writes `torch.tensor(ids, dtype=torch.long)` for exactly this reason.

👉 **Predict first:** which dtype does each tensor get?

````python
token_ids = torch.tensor([41, 40], dtype=torch.long)
weights = torch.tensor([0.5, -1.2])
print(token_ids.dtype)
print(weights.dtype)
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`torch.int64` (that's what `long` means) and `torch.float32`. PyTorch picks decimals for numbers with a point.

</details>

### ❓ How do I take a piece of a tensor?

With a slice, exactly like a Python list: `[start:end]`. The training loop uses `data[i : i + chunk_size]` to cut a window of 24 tokens from the book.

👉 **Predict first:** what's `data[1:4]`?

````python
data = torch.tensor([10, 11, 12, 13, 14, 15])
print(data[1:4])
print(data[0].item())
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`tensor([11, 12, 13])`, then `10`. `.item()` turns a one-number tensor back into a plain Python number.

</details>

### ❓ Why does the code set `torch.manual_seed(42)`?

A model starts with random weights. Computers make "random" numbers from a starting number, the **seed**. Same seed, same numbers, every run. That's why our training gives the same loss every time, so experiments can be compared.

👉 **Predict first:** will the two lines be the same?

````python
torch.manual_seed(42)
print(torch.rand(3))
torch.manual_seed(42)
print(torch.rand(3))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

The same three "random" numbers twice. `torch.rand(3)` makes 3 random decimals between 0 and 1.

</details>

## 5. Full example

The real Book 1 of the Odyssey, as one tensor, the way `train_homer.py` builds it.

````python
from homer_ai.train_homer import BOOKS_DIR, load_data

data, tokenizer, text = load_data(BOOKS_DIR)
print(data.shape, data.dtype)
print(data[:10])
print(tokenizer.decode(data[:10].tolist()))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

One long row of 19,701 whole numbers, one per character. The first 10 decode back to the opening of the poem. `.tolist()` turns a tensor back into a plain Python list.

</details>
