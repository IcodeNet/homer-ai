### ❓ How do I store a value?

Give it a name with `=`. The name is a **variable**: a label pointing at a value. `=` here means "let this name point to", not "is equal to".

Text goes between quotes, and is called a **string**.

👉 **Predict first:** what will `print` show?

````python
word = "μῆνιν"
print(word)
print(len(word))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

The word, then `5`. `len` counts how many characters a string has.

</details>

### ❓ How do I keep several values together?

In a **list**: values in square brackets, separated by commas. Positions start at **0**, so `[0]` is the first one.

👉 **Predict first:** what is `letters[0]`, and what is `letters[-1]`?

````python
letters = ["μ", "ῆ", "ν", "ι", "ν"]
print(letters[0])
print(letters[-1])
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`μ` and `ν`. A negative position counts from the end: `-1` is the last item.

</details>

### ❓ How do I take a piece of a string or list?

With a **slice**: `[start:end]`. It takes from `start` up to, but **not** including, `end`. A blank start means "from the beginning"; a blank end means "to the end". The training code uses this to cut the text into pieces.

👉 **Predict first:** what is `word[1:3]`?

````python
word = "μῆνιν"
print(word[1:3])
print(word[:2], word[2:])
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`ῆν`, then `μῆ νιν`. Positions 1 and 2, then the word split at position 2.

</details>

### ❓ What's the difference between `append` and `extend`?

Both add to a list. `append` adds **one** item, even if that item is itself a list. `extend` pours the items in **one by one**.

👉 **Predict first:** what do `x` and `y` look like?

````python
x = [1]
x.append([2, 3])
y = [1]
y.extend([2, 3])
print(x)
print(y)
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`[1, [2, 3]]` (a box inside the box) and `[1, 2, 3]`. `load_data` uses `extend` to add a whole book of numbers.

</details>

### ❓ How do I look something up?

With a **dictionary**: pairs of `key: value` in curly brackets. Square brackets look up a key, and crash with `KeyError` if it's missing. `.get(key, backup)` gives the backup value instead. Our tokenizer uses this for unknown characters.

👉 **Predict first:** what does `.get("ω", 99)` give?

````python
char_to_id = {"α": 0, "β": 1}
print(char_to_id["α"])
print(char_to_id.get("ω", 99))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`0`, then `99`, because `ω` isn't in the dictionary.

</details>

### ❓ How do I repeat something for every item?

With a **for loop**. The indented lines run once for each item. **Indentation** (the spaces at the start of a line) shows which lines belong to the loop.

`if` runs its indented lines only when something is true.

👉 **Predict first:** which letters get printed?

````python
for letter in "μῆνιν":
    if letter == "ν":
        print("found ν")
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`found ν` twice, because `ν` appears twice. `==` means "is equal to" (one `=` gives a name; two compare).

</details>

### ❓ What is a list comprehension?

A short way to build a list from a loop, in one line: `[what_to_keep for item in things]`. Our tokenizer's `encode` is one.

👉 **Predict first:** what list do we get?

````python
char_to_id = {"μ": 0, "ῆ": 1, "ν": 2, "ι": 3}
[char_to_id[c] for c in "μῆνιν"]
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`[0, 1, 2, 3, 2]`: one number per letter. That's exactly what encoding text means.

</details>

### ❓ What is a function, and why didn't my `print` inside it run?

A **function** is a named block of code. `def` only **defines** it, like building a machine. Nothing happens until you **call** it, with brackets, like switching the machine on. `return` sends a value back to whoever called it.

👉 **Predict first:** does anything print when the `def` block runs?

````python
def greet(name):
    return "Χαῖρε " + name


print("defined, not called yet")
print(greet("Homer"))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

The `def` printed nothing. Only `greet("Homer")`, the call, ran the code inside.

</details>

### ❓ What is an f-string?

A string with `f` in front. Anything in `{curly brackets}` is replaced by its value. The training loop prints its progress this way.

👉 **Predict first:** what is printed?

````python
epoch = 50
loss = 2.38123
print(f"Epoch {epoch} | loss {loss:.2f}")
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`Epoch 50 | loss 2.38`. `:.2f` means "show 2 decimal places".

</details>

### ❓ What is a class?

A **class** is a blueprint for making objects that keep their own data and functions together. `CharTokenizer` is a class: each tokenizer you make from it remembers its own vocabulary. A function that belongs to a class is called a **method**, like `encode`.

👉 **Predict first:** what does `t.vocab_size` give for the text `"abc"`?

````python
from homer_ai.tokenizer import CharTokenizer

t = CharTokenizer("abc")
print(t.vocab_size)
print(t.encode("ab"))
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

`5`: the 3 letters plus our 2 special tokens. Then `[0, 1]`. `from … import …` loaded the class from our own file, `tokenizer.py`.

</details>

## 5. Full example

Everything together: turn a Greek line into numbers and back, with only the ideas from this notebook.

````python
line = "μῆνιν ἄειδε θεὰ"
chars = sorted(set(line))  # every different character, in order
char_to_id = {ch: i for i, ch in enumerate(chars)}
id_to_char = {i: ch for ch, i in char_to_id.items()}

ids = [char_to_id[c] for c in line]
back = "".join(id_to_char[i] for i in ids)

print(f"{len(chars)} different characters")
print(ids)
print(back == line)
````

<details><summary>👀 What you saw (open it after you've run the cell)</summary>

The numbers for each character, and `True`: decoding gives back the same line. (`set` keeps one of each, `sorted` puts them in order, `enumerate` numbers them 0, 1, 2…, and `"".join` glues characters into a string.)

</details>
