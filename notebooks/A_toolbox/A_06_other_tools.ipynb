## 4. Step by step

These tools run in the terminal, so this notebook shows the commands rather than running them.

### ❓ What does uv do?

**uv** installs Python and every library the project needs, and runs your code with them. The three commands you'll use:

```
uv sync                 # install everything on the shopping list
uv run python file.py   # run a file with the project's Python
uv add numpy            # add a new library to the list
```

`pyproject.toml` is the shopping list, and `uv.lock` records the exact versions, so everyone gets the same ones.

### ❓ What does Ruff do?

**Ruff** is a linter. It reads your code and points out mistakes without running it, like a spell-checker for code. It also formats code (fixes spacing and layout). In this project VS Code runs it every time you save.

```
uv run ruff check src/      # look for mistakes
uv run ruff format src/     # tidy the layout
```

Each problem has a code, like `F821` ("undefined name": you used a name that doesn't exist) or `E501` ("line too long").

### ❓ How do I silence one Ruff warning?

Add `# noqa: E501` at the end of that line, with the rule's code. Always **name the rule**. A bare `# noqa` silences *every* rule on the line, so a real mistake there would be hidden too. Silence the smallest thing possible.

### ❓ How do notebooks run Python, and why do we need `ipykernel`?

A notebook doesn't run Python itself. It sends each cell to the **kernel**, a Python program running in the background, and shows the result. `ipykernel` is the helper that turns Python into a kernel. It's on the shopping list (`pyproject.toml`), so `uv sync` installs it for every new developer. Choose the kernel in `.venv` (top right: Select Kernel).

### ❓ Should I run one cell or run all?

Shift+Enter (one cell) while you're learning, so you can predict each result. Run All after reopening a notebook, because the kernel forgets everything when it restarts.

### ❓ My notebook opens as JSON. Why?

A notebook is a **JSON** file underneath (plain text that stores data as names and values), and VS Code opened it with the text editor. Right-click it → Open With… → Jupyter Notebook.

### ❓ What are git commits and branches?

**git** saves snapshots of the project. A commit is one snapshot, like a save point in a game, and you can always go back to it. A branch is a separate line of work, so the main version stays safe while you try things.

```
git status                  # what has changed?
git add file.py             # choose what goes in the next snapshot
git commit -m "message"     # save the snapshot
git switch -c my-branch     # start a new branch
```

### ❓ What's MCP, and how does our server use it?

**MCP** (Model Context Protocol) is a standard way for AI apps like Claude to call outside tools. `homer_mcp_server.py` loads our trained model and offers it as a tool, `consult_homer_llm`. Give it some Greek, and it writes a continuation.

A line starting with `@` above a function, like `@mcp.tool()`, is a **decorator**. It wraps the function. Here it registers the function as a tool that Claude can see and call.

## 5. Full example: a working day

```
uv sync                                      # 1. set up (once)
uv run homer-train --model gpt               # 2. train the GPT (about 3 minutes)
uv run ruff check src/                       # 3. check the code
git add src/ && git commit -m "Try lr 0.003"  # 4. save a snapshot
```

Then any MCP app, like Claude, can use the trained model through `homer_mcp_server.py`.
